````python
import requests

def generate_response(prompt, model="llama3.2"):
    try:
        response = requests.post(
            "http://localhost:11434/api/generate",
            json={
                "model": model,
                "prompt": prompt,
                "stream": False
            }
        )
        response.raise_for_status() # Lanza un error para códigos de estado HTTP erróneos
        return response.json()["response"]
    except requests.exceptions.RequestException as e:
        return f"Error al conectar con Ollama o generar respuesta: {e}"


# Prueba la generación de respuesta
ollama_response = generate_response("Explica brevemente qué es la inteligencia artificial.")
print("\nRespuesta de Llama 3.2 a través de Ollama:")
print(ollama_response)
````

````python
template = """Eres un experto en análisis de complejidad algorítmica.
Analiza el siguiente código y proporciona su complejidad algorítmica en notación Big O.
Código:
```python
{code}```
Tu tarea es tomar el siguiente codigo desarrollado en cualquier lenguaje de programacion y crear una variable
alterna del mismo codigo agregando comentarios. Estos comentarios deberan contener la complejidad algoritmica en notacion big-0
linea por linea. Ademas, debes seguir el estilo de comentarios que correspondan al lenguaje de programacion.  Añade un bloque al inicio, donde se describa el proposito general del codigo
y añade un bloque final donde se describa la complejidad algorítmica total del código.
"""
````

````python
def lectura_archivo(ruta_archivo):
    with open(ruta_archivo, 'r') as file:
        return file.read()
````

````python
import os
````

````python
def guardar_resultado(ruta_salida, resultado):
    nombre_base, extension = os.path.splitext(ruta_salida)
    ruta_comentada = f"{nombre_base}_comentado{extension}"
    with open(ruta_comentada, 'w') as file:
        file.write(resultado)
    return ruta_comentada
````

````python
def analizar_complejidad(codigo):
    return generate_response(template.format(code=codigo))
````

````python
def inicio_analisis():
    print("Iniciando análisis de complejidad algorítmica...")
    # ruta_archivo = input("Indique la ruta del archivo a analizar: ")
    ruta_archivo = "clases-1.cpp"
    codigo = lectura_archivo(ruta_archivo)
    print(f"Accediento y analizando el archivo: {ruta_archivo}")
    try:
        respuesta = analizar_complejidad(codigo)
        print("Análisis completado. Guardando resultado...")
        ruta_salida = guardar_resultado(ruta_archivo, respuesta)
        print(f"Resultado guardado en: {ruta_salida}")
        print("--- fragmento de codigo ---")
        print(respuesta[:500])  # Print first 500 characters of the result
    except Exception as e:
        print(f"Error durante el análisis: {e}")  
````

````python
inicio_analisis()
````
